### Missing Data Handling

The missing-value analysis identified missing data in the Age, Cabin, and Embarked columns. The Age column contains 177 missing values (19.87%), so median imputation is used because Age is a numerical variable and the median is less sensitive to extreme values. The Cabin column contains 687 missing values (77.10%), which is a very high proportion; therefore, the column is removed because retaining it would provide limited reliable information. The Embarked column contains only 2 missing values (0.22%), so mode imputation is applied because it is a categorical variable with very few missing observations.

```python
# Step 5: Missing Data Handling

# Age: Fill missing values with median
age_median = df['Age'].median()
df['Age'] = df['Age'].fillna(age_median)

# Embarked: Fill missing values with mode
embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

# Cabin: Drop because 77.10% of values are missing
df = df.drop(columns=['Cabin'])

print("Missing data handling completed successfully!")
print("Age median used:", age_median)
print("Embarked mode used:", embarked_mode)
print("Cabin column removed due to high missing percentage.")

print("\nMissing values after handling:")
display(df.isnull().sum())
```

```python
# Step 6: Duplicate Row Removal

duplicates_before = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)

df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicates_after)
print("Duplicate removal completed successfully!")
print("Dataset shape after duplicate removal:", df.shape)
```

### Data Standardisation

The categorical values were checked for inconsistent formatting. The Sex column was standardised to lowercase and then converted to a consistent format. The Embarked column was also checked for unique categorical values. Identifier fields such as PassengerId and Ticket were retained as identifiers rather than treated as numerical measurements.

```python
# Step 7: Data Standardisation

# Standardise Sex values
df['Sex'] = df['Sex'].astype(str).str.strip().str.lower()

# Standardise Embarked values
df['Embarked'] = df['Embarked'].astype(str).str.strip().str.upper()

# Convert PassengerId to string because it is an identifier
df['PassengerId'] = df['PassengerId'].astype(str)

print("Data standardisation completed successfully!")

print("\nUnique Sex values:")
print(df['Sex'].unique())

print("\nUnique Embarked values:")
print(df['Embarked'].unique())

print("\nUpdated data types:")
display(df.dtypes)
```

### Outlier Detection Using IQR

The Interquartile Range (IQR) method was used to identify potential outliers in the numerical variables. The IQR is calculated as the difference between the third quartile (Q3) and the first quartile (Q1). Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are identified as potential outliers. The analysis is performed on Age and Fare, while identifier and categorical variables are excluded.

```python
# Step 8: Outlier Detection Using IQR

numeric_columns = ['Age', 'Fare']

outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        'Column': column,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_report = pd.DataFrame(outlier_summary)

print("Outlier detection completed successfully!")
display(outlier_report)
```

### Outlier Treatment Decision

The IQR method identified 66 potential outliers in Age and 116 potential outliers in Fare. These observations were retained because the extreme values may represent genuine passenger characteristics, such as older passengers or passengers who paid higher fares. Removing them could result in loss of meaningful information. Therefore, the identified outliers were documented but not removed from the cleaned dataset.
